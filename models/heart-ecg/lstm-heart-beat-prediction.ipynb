# Task
Build and train an LSTM model for classification using the `mitbih_train.csv` dataset, with column 187 as the target variable.

## Analyze the target variable

### Subtask:
Analyze the distribution of the target variable (column 187) to understand the class balance.


**Reasoning**:
To analyze the distribution of the target variable, I will get the value counts and percentages of column 187 in the `mitbih_train_df` DataFrame.



```python
target_variable_counts = mitbih_train_df[187].value_counts()
print("Count of each class in the target variable (column 187):")
print(target_variable_counts)

target_variable_percentages = mitbih_train_df[187].value_counts(normalize=True) * 100
print("\nPercentage distribution of each class in the target variable (column 187):")
print(target_variable_percentages)
```

## Prepare data for LSTM

### Subtask:
Separate features (X) and target (y), reshape X for LSTM input, and one-hot encode the target variable.


**Reasoning**:
To prepare the data for an LSTM model, I will separate features from the target, reshape the features into a 3D array for LSTM input, and then one-hot encode the target variable.



```python
import numpy as np
from tensorflow.keras.utils import to_categorical

# 1. Separate features (X) and target (y)
X = mitbih_train_df.iloc[:, :187]
y = mitbih_train_df.iloc[:, 187]

# 2. Reshape X for LSTM input
X = X.values.reshape(X.shape[0], X.shape[1], 1)

# 3. One-hot encode the target variable
y_encoded = to_categorical(y)

print("Shape of X (features):")
print(X.shape)
print("Shape of y (target):")
print(y.shape)
print("Shape of y_encoded (one-hot encoded target):")
print(y_encoded.shape)
```

## Split data into training and validation sets

### Subtask:
Divide the prepared data into training and validation sets for model training and evaluation.


**Reasoning**:
To divide the data into training and validation sets, I will import `train_test_split` and use it with the specified parameters: an 80/20 ratio, `random_state=42`, and stratification on the original target variable `y`.



```python
from sklearn.model_selection import train_test_split

# Split data into training and validation sets (80/20 ratio) with stratification
X_train, X_val, y_train, y_val = train_test_split(X, y_encoded, test_size=0.2, random_state=42, stratify=y)

# Print the shapes of the resulting datasets
print("Shape of X_train:", X_train.shape)
print("Shape of X_val:", X_val.shape)
print("Shape of y_train:", y_train.shape)
print("Shape of y_val:", y_val.shape)
```

## Build and compile LSTM model

### Subtask:
Define the architecture of the Keras LSTM model and compile it with appropriate optimizer, loss, and metrics.


**Reasoning**:
To build the LSTM model, I need to import the required layers from Keras and define the model architecture as specified, including LSTM, Dropout, and Dense layers, then compile it with the appropriate optimizer, loss function, and metrics.



```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout

# 1. Initialize a Sequential model
model = Sequential()

# 2. Add an LSTM layer
# X_train.shape[1] is the number of time steps (187 features)
# X_train.shape[2] is the number of features per time step (1, as it's a univariate series)
model.add(LSTM(units=100, return_sequences=False, input_shape=(X_train.shape[1], X_train.shape[2])))

# 3. Add a Dropout layer
model.add(Dropout(0.2))

# 4. Add a Dense output layer
# y_encoded.shape[1] gives the number of classes (5)
model.add(Dense(units=y_encoded.shape[1], activation='softmax'))

# 5. Compile the model
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

print("LSTM model architecture:")
model.summary()
```

**Reasoning**:
Now that the LSTM model architecture is defined and compiled, the next logical step is to train the model using the prepared training and validation datasets.



```python
history = model.fit(X_train, y_train, epochs=20, batch_size=32, validation_data=(X_val, y_val))

print("Model training complete.")
```

# Task
Evaluate the trained LSTM model on the validation set using `X_val` and `y_val` to determine its accuracy and loss, then summarize the model's performance and provide insights from the evaluation.

## Evaluate LSTM model

### Subtask:
Evaluate the trained LSTM model on the validation set to determine its accuracy and loss. This will involve using the `model.evaluate()` method with `X_val` and `y_val`.


**Reasoning**:
To evaluate the trained LSTM model, I will use the `model.evaluate()` method with the validation datasets (`X_val` and `y_val`) and then print the resulting loss and accuracy.



```python
loss, accuracy = model.evaluate(X_val, y_val, verbose=0)
print(f"Validation Loss: {loss:.4f}")
print(f"Validation Accuracy: {accuracy:.4f}")
```

## Save model for frontend/API use

Save the trained LSTM model and metadata so the backend can load it (e.g. for an ECG heartbeat prediction endpoint).

```python
import json

# Output directory for frontend/API (run notebook from heart-ecg folder -> saved_model/ here)
SAVED_MODEL_DIR = os.path.join(os.getcwd(), "saved_model")
os.makedirs(SAVED_MODEL_DIR, exist_ok=True)

# 1. Save Keras model (load with tf.keras.models.load_model)
model_path = os.path.join(SAVED_MODEL_DIR, "lstm_heartbeat.keras")
model.save(model_path)
print(f"Model saved to: {model_path}")

# 2. Save metadata for frontend/API (input shape, classes, metrics)
# MIT-BIH class mapping: 0=Normal, 1=Supraventricular, 2=Premature ventricular, 3=Fusion, 4=Unknown
class_names = ["Normal", "Supraventricular premature beat", "Premature ventricular contraction", "Fusion of normal and PVC", "Unclassified"]
metadata = {
    "model_name": "LSTM Heartbeat (MIT-BIH)",
    "model_type": "keras_lstm",
    "model_file": "lstm_heartbeat.keras",
    "input_shape": [int(X_train.shape[1]), int(X_train.shape[2])],
    "sequence_length": int(X_train.shape[1]),
    "num_classes": int(y_encoded.shape[1]),
    "class_names": class_names,
    "accuracy": float(accuracy),
    "loss": float(loss),
}
metadata_path = os.path.join(SAVED_MODEL_DIR, "model_metadata.json")
with open(metadata_path, "w") as f:
    json.dump(metadata, f, indent=2)
print(f"Metadata saved to: {metadata_path}")
print("Done. Backend can load model from:", os.path.abspath(SAVED_MODEL_DIR))
```
